### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
#check for actual duplicate response IDs
print(df["ResponseId"].duplicated().sum())

#identify repeated Response IDs
print(df[df.duplicated(subset=["ResponseId"], keep=False)])
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
#making a copy of original df before making changes
df_clean = df.copy()

#ensuring the the copied df is providing the same information
#print(df_clean["ResponseId"].duplicated().sum())
#print(df_clean[df.duplicated(subset=["ResponseId"], keep=False)])

#delete those duplicate rows
df_clean = df_clean.drop_duplicates(subset=["ResponseId"])

#compare the numbner of rows
print("Original rows:", len(df))
print("Clean rows:", len(df_clean))

#verify that no duplicates remain in the copied df
print(df_clean["ResponseId"].duplicated().sum())
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here

#Displays all the columns
pd.set_option("display.max_rows", None)

#counts missing values in each column and sort them 
print(df_clean.isnull().sum().sort_values(ascending=False))

#Find the most frequent value in EdLevel and then impute it in all the missing values in that column

Most_EdLevel = df_clean["EdLevel"].mode()[0]
print("Most frequent Ed Level is", Most_EdLevel)

#Impute with most frequent Edlevel

df_clean["EdLevel"] = df_clean["EdLevel"].fillna(Most_EdLevel)
print(df_clean["EdLevel"].head())
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
## Write your code here

#count the total number of missing values in the ConvertedCompYearly column

print(df_clean["ConvertedCompYearly"].isnull().sum())

#Find most common compensation amount 

Most_Comp = df_clean["ConvertedCompYearly"].mode()[0]
print("most frequent value for compensation is $",Most_Comp)

#Impute with most frequent Compensation amount
df_clean["ConvertedCompYearly"] = df_clean["ConvertedCompYearly"].fillna(Most_Comp)
print(df_clean["ConvertedCompYearly"].head())

#Check to make sure EdLevel and ConvertedCompYearly have no missign values 

print(df_clean.isnull().sum().sort_values(ascending=False))
```
